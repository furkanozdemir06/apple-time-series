```python
model=sm.tsa.statespace.SARIMAX(df['Close'])
result=model.fit()
```

```python
result.summary()
```

According to the results of the SARIMAX(1, 0, 0) model trained on 494 observations, the AR(1) coefficient was calculated as $0.9999$ and found to be statistically significant ($p = 0.000$). The Ljung-Box test ($Prob(Q) = 0.16 > 0.05$) and the Heteroskedasticity test ($Prob(H) = 0.38 > 0.05$) indicate that the residuals exhibit white noise properties and that the model successfully captures the autocorrelation structure. However, the Jarque-Bera test ($Prob(JB) = 0.00 < 0.05$) alongside a high kurtosis value ($Kurtosis = 9.31$) shows that the model residuals do not follow a normal distribution.

```python
# %% Train / test split
import numpy as np
import pandas as pd
import plotly.graph_objects as go
from sklearn.metrics import mean_absolute_error, mean_squared_error
 
close = df['Close']
test_size = 30
 
train = close.iloc[:-test_size]
test = close.iloc[-test_size:]
 
print("Train:", train.shape, "| Test:", test.shape)
```

```python
# %% Fit on train and forecast the test period
# order=(1,1,1): d=1 differences the series, since prices are non-stationary
model = sm.tsa.statespace.SARIMAX(train, order=(1, 1, 1))
result = model.fit(disp=False)
 
fc = result.get_forecast(steps=test_size)
pred = pd.Series(np.asarray(fc.predicted_mean), index=test.index)
conf = pd.DataFrame(np.asarray(fc.conf_int()), index=test.index, columns=['lower', 'upper'])
```

```python
# %% Naive baseline: "tomorrow = last known price"
naive = pd.Series(train.iloc[-1], index=test.index)
```

```python
# %% Evaluation
def evaluate(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mape = np.mean(np.abs((y_true - y_pred) / y_true)) * 100
    return mae, rmse, mape
 
scores = pd.DataFrame(
    [evaluate(test, pred), evaluate(test, naive)],
    columns=['MAE', 'RMSE', 'MAPE (%)'],
    index=['SARIMAX(1,1,1)', 'Naive baseline']
).round(3)
 
scores
```

```python
# %% Plot: actual vs forecast
fig = go.Figure()
fig.add_trace(go.Scatter(x=train.index[-90:], y=train.iloc[-90:], name='Train (last 90 days)'))
fig.add_trace(go.Scatter(x=test.index, y=test, name='Actual'))
fig.add_trace(go.Scatter(x=pred.index, y=pred, name='SARIMAX forecast'))
fig.add_trace(go.Scatter(x=naive.index, y=naive, name='Naive baseline', line=dict(dash='dash')))
fig.add_trace(go.Scatter(
    x=list(conf.index) + list(conf.index[::-1]),
    y=list(conf['upper']) + list(conf['lower'][::-1]),
    fill='toself', fillcolor='rgba(0,100,200,0.15)',
    line=dict(color='rgba(255,255,255,0)'), name='95% confidence interval'
))
fig.update_layout(title='AAPL Close: Forecast vs Actual', xaxis_title='Date', yaxis_title='Price (USD)')
fig.show()
```

```python
# %% Future forecast: refit on all data, predict the next 30 business days
final_model = sm.tsa.statespace.SARIMAX(close, order=(1, 1, 1)).fit(disp=False)
future = final_model.get_forecast(steps=30)
 
future_dates = pd.bdate_range(start=close.index[-1] + pd.Timedelta(days=1), periods=30)
future_pred = pd.Series(np.asarray(future.predicted_mean), index=future_dates)
future_conf = pd.DataFrame(np.asarray(future.conf_int()), index=future_dates, columns=['lower', 'upper'])
 
fig = go.Figure()
fig.add_trace(go.Scatter(x=close.index[-120:], y=close.iloc[-120:], name='History'))
fig.add_trace(go.Scatter(x=future_pred.index, y=future_pred, name='Forecast'))
fig.add_trace(go.Scatter(
    x=list(future_conf.index) + list(future_conf.index[::-1]),
    y=list(future_conf['upper']) + list(future_conf['lower'][::-1]),
    fill='toself', fillcolor='rgba(0,100,200,0.15)',
    line=dict(color='rgba(255,255,255,0)'), name='95% confidence interval'
))
fig.update_layout(title='AAPL Close: 30-Day Forecast', xaxis_title='Date', yaxis_title='Price (USD)')
fig.show()
```
